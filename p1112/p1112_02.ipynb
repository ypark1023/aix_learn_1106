```python
# 그래프 관련
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np

# 머신러닝 관련
from sklearn import svm,metrics
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import StandardScaler, PolynomialFeatures, OneHotEncoder
from sklearn.metrics import mean_absolute_error 
```

```python
df = pd.read_csv('data/manhattan.csv')
df.head(3)
```

```python
# data = df.iloc[:,2:16]
# label = df['rent']
# data = np.array(data)
# data
label = df['rent']
data = df[df.columns[2:-2]].values
```

```python
train_data,test_data,train_label,test_label = train_test_split(data,label,random_state=3539)
```

```python
# # 1차원
# lr = LinearRegression()
# lr.fit(train_data, train_label)
# predict = lr.predict(test_data)
# # print('점수 예측: ', predict)

# score = lr.score(train_data, train_label)
# score2 = lr.score(test_data, test_label)
# print(f'정확도:{score*100}%, {score2*100}%')
```

```python
poly = PolynomialFeatures(degree=2,include_bias=False)  #include_bias=False 학습의미없는 데이터 1를 생략
poly.fit(train_data)
train_poly = poly.transform(train_data)
test_poly = poly.transform(test_data)
# train_poly[:2]
```

```python
# 2차원
lr2 = LinearRegression()
lr2.fit(train_poly, train_label)
predict2 = lr2.predict(test_poly)
# print('점수 예측: ', predict)

score3 = lr2.score(train_poly, train_label)
score4 = lr2.score(test_poly, test_label)
print(f'정확도:{score3*100}%, {score4*100}%')
```

```python
# StandartScaler
ss = StandardScaler()
ss.fit(train_poly)
train_scaled = ss.transform(train_poly)
test_scaled = ss.transform(test_poly)
```

```python
# 규제 알고리즘

ridge = Ridge(alpha=1)
ridge.fit(train_scaled, train_label)
score5 = ridge.score(train_scaled, train_label)
score6 = ridge.score(test_scaled, test_label)
print(f'정확도:{score5*100}%, {score6*100}%')
```
