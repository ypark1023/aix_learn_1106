```python
# 그래프 관련
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np

# 머신러닝 관련
from sklearn import svm,metrics
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_validate
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.preprocessing import StandardScaler, PolynomialFeatures, OneHotEncoder
from sklearn.metrics import mean_absolute_error 
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from scipy.special import expit
from scipy.special import softmax
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import RandomForestRegressor
```

```python
df = pd.read_csv('data/picher_stats_2017.csv')
df.head(3)
```

```python
df.columns
```

```python
label = df['연봉(2018)']
data = df[df.columns[2:-2]]
data.head(2)
```

```python
train_data,test_data,train_label,test_label = train_test_split(data,label,test_size=0.2,random_state=42)
```

```python
# 2. 랜덤 포레스트 모델 학습 (n_estimators: 나무 개수, random_state: 고정)
rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(train_data, train_label)

# 3. 정확도($R^2$ Score) 출력
print(f'학습 세트 정확도($R^2$): {rf.score(train_data, train_label)*100:.2f}%')
print(f'테스트 세트 정확도($R^2$): {rf.score(test_data, test_label)*100:.2f}%')

# 4. 새로운 단일 데이터 예측 (폴리/스케일링 과정 불필요!)
input1 = [9, 8, 0, 2, 0, 35, 24, 137.1, 9.11, 4.19, 0.52, 0.321, 73.1, 3.67, 4.99, 3.91, 3.67, 3.78]

# scikit-learn 입력 규격(2차원 배열)에 맞게 변환
# 주의: 모델이 학습할 때 사용한 컬럼 이름(열 이름) 순서와 일치해야 가장 정확합니다.
input_data = np.array(input1).reshape(1, -1)

ipredict = rf.predict(input_data)
print('연봉 예측: ', ipredict[0])
```
