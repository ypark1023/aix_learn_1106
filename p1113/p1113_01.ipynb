```python
# 그래프 관련
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np

# 머신러닝 관련
from sklearn import svm,metrics
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import StandardScaler, PolynomialFeatures, OneHotEncoder
from sklearn.metrics import mean_absolute_error 
from sklearn.linear_model import LogisticRegression
from scipy.special import expit
from scipy.special import softmax
```

```python
fish = pd.read_csv('data/fish_logistic.csv')
fish.head(2)
```

```python
data = fish[fish.columns[1:]].values
label = fish['Species'].values
```

```python
train_data,test_data,train_label,test_label = train_test_split(data,label)
train_data[:2]
```

```python
# StandartScaler 정규화
ss = StandardScaler()
ss.fit(train_data)
train_scaled = ss.transform(train_data)
test_scaled = ss.transform(test_data)
```

```python
# 데이터 학습
lr = LogisticRegression(C=20, max_iter=1000)
lr.fit(train_scaled, train_label)
predict = lr.predict(test_scaled)
print(f'학습 정확도: {lr.score(train_scaled, train_label)*100}%')
print(f'테스트 정확도: {lr.score(test_scaled, test_label)*100}%')
```

```python
predict = lr.predict(test_scaled[:5])
predict
```

```python
# 예측 시 예측확률 보여줌
decision = lr.decision_function(test_scaled[:5])
soft1 = softmax(decision, axis=1)
# decision
# soft1
print(lr.classes_)
np.round(soft1,decimals=3)
```
